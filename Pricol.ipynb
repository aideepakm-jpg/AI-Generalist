# 1..Import Libraries

```python
import pandas as pd
from pycaret.classification import *
```

# 2.Load Dataset

```python
import pandas as pd
df=pd.read_csv('pricol.csv')
df.head()
```

```python
df = df.drop(['Unnamed: 0'], axis=1)
```

```python
print(df.info())
```

# 3.Checking Missing Values


```python
print(df.isnull().sum())
```

# 4.Create target column

```python
df['Target'] = (df['close'].shift(-1) > df['close']).astype(int)
```

```python
df = df.drop(['ticker'], axis=1)
```

# 5.Fill the missing values

```python
df = df.ffill()
```

```python
print(df.isnull().sum())
```

```python
df.head()
```

# 6.Set up Pycaret

```python
s = setup(
    data=df,
    target='Target',
    session_id=123,
    normalize=True
)
```

# 7.Compare Algorithms


```python
best_model = compare_models()
```

# 8.Train Best Model

```python
final_model = finalize_model(best_model)
```

# 9.Check Accuracy

```python
evaluate_model(final_model)
```

# 10.Predict

```python
predictions = predict_model(final_model)
print(predictions.head())
```
